```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("forestfires.csv")
```

```python
# Set style visualisasi
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
```

1. Pada bulan apa kejadian kebakaran hutan paling sering terjadi?


```python
plt.figure()
sns.countplot(data=df, x='month', order=['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec'])
plt.title("Jumlah Kebakaran Hutan per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah Kejadian")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

2. Hari apa yang paling sering terjadi kebakaran hutan?


```python
plt.figure()
sns.countplot(data=df, x='day', order=['mon','tue','wed','thu','fri','sat','sun'])
plt.title("Jumlah Kebakaran Hutan per Hari")
plt.xlabel("Hari")
plt.ylabel("Jumlah Kejadian")
plt.tight_layout()
plt.show()
```

3. Bagaimana distribusi suhu (temp) saat terjadi kebakaran hutan?

```python
plt.figure()
sns.histplot(df['temp'], bins=15, kde=True)
plt.title("Distribusi Suhu saat Kebakaran")
plt.xlabel("Suhu (°C)")
plt.ylabel("Frekuensi")
plt.tight_layout()
plt.show()
```

4. Apakah ada hubungan antara suhu (temp) dan luas area terbakar (area)?


```python
plt.figure()
sns.scatterplot(data=df, x='temp', y='area')
plt.title("Hubungan Suhu dan Luas Area Terbakar")
plt.xlabel("Suhu (°C)")
plt.ylabel("Luas Area Terbakar (ha)")
plt.tight_layout()
plt.show()
```

5. Seberapa besar peran curah hujan (rain) dalam mengurangi luas kebakaran (area)?

```python
plt.figure()
sns.scatterplot(data=df, x='rain', y='area')
plt.title("Pengaruh Curah Hujan terhadap Luas Area Kebakaran")
plt.xlabel("Curah Hujan (mm)")
plt.ylabel("Luas Area Terbakar (ha)")
plt.tight_layout()
plt.show()
```

6. Bagaimana hubungan antara kecepatan angin (wind) dan luas kebakaran (area)?


```python
plt.figure()
sns.scatterplot(data=df, x='wind', y='area')
plt.title("Hubungan Kecepatan Angin dan Luas Area Terbakar")
plt.xlabel("Kecepatan Angin (km/h)")
plt.ylabel("Luas Area Terbakar (ha)")
plt.tight_layout()
plt.show()
```

7. berapa luas area hutan yang terbakar setiap bulannya?

```python
plt.figure()
total_area_by_month = df.groupby('month')['area'].sum().reindex(['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec'])
sns.barplot(x=total_area_by_month.index, y=total_area_by_month.values)
plt.title("Total Luas Area Hutan yang Terbakar per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Total Area Terbakar (ha)")
plt.tight_layout()
plt.show()
```

8. Distribusi nilai kelembaban relatif (RH) saat terjadi kebakaran seperti apa?


```python
plt.figure()
sns.histplot(df['RH'], bins=20, kde=True)
plt.title("Distribusi Kelembaban Relatif saat Kebakaran")
plt.xlabel("Kelembaban Relatif (%)")
plt.ylabel("Frekuensi")
plt.tight_layout()
plt.show()
```

9. Seberapa sering terjadi kebakaran di setiap kombinasi koordinat lokasi (X, Y)?


```python
plt.figure()
heatmap_data = df.groupby(['Y', 'X']).size().unstack(fill_value=0)
sns.heatmap(heatmap_data, cmap="YlOrRd", annot=True, fmt="d")
plt.title("Heatmap Lokasi Kebakaran (Koordinat X, Y)")
plt.xlabel("X")
plt.ylabel("Y")
plt.tight_layout()
plt.show()
```

10. Bagaimana tren indeks kekeringan (FFMC, DMC, DC) terhadap luas area yang terbakar?


```python
fig, axes = plt.subplots(3, 1, figsize=(10, 18))

sns.scatterplot(data=df, x='FFMC', y='area', ax=axes[0])
axes[0].set_title("FFMC vs Luas Area Terbakar")
axes[0].set_xlabel("FFMC")
axes[0].set_ylabel("Area (ha)")

sns.scatterplot(data=df, x='DMC', y='area', ax=axes[1])
axes[1].set_title("DMC vs Luas Area Terbakar")
axes[1].set_xlabel("DMC")
axes[1].set_ylabel("Area (ha)")

sns.scatterplot(data=df, x='DC', y='area', ax=axes[2])
axes[2].set_title("DC vs Luas Area Terbakar")
axes[2].set_xlabel("DC")
axes[2].set_ylabel("Area (ha)")

plt.tight_layout()
plt.show()
```
